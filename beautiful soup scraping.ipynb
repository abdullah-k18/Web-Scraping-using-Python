```python
from bs4 import BeautifulSoup
import requests

main_url = ""

main_page = requests.get(main_url)
main_soup = BeautifulSoup(main_page.text, "html.parser")

iframe_tags = main_soup.find_all("iframe", title="Embedded post")

for idx, iframe_tag in enumerate(iframe_tags, start=1):
    iframe_src = iframe_tag.get("src")

    if iframe_src:
        iframe_page = requests.get(iframe_src)
        iframe_soup = BeautifulSoup(iframe_page.text, "html.parser")

        div_tags = iframe_soup.find_all("div", class_="attributed-text-segment-list__container relative mt-1 mb-1.5 babybear:mt-0 babybear:mb-0.5")

        for div_idx, div_tag in enumerate(div_tags, start=1):
            p_tag = div_tag.find("p", class_="attributed-text-segment-list__content text-color-text !text-sm whitespace-pre-wrap break-words")

            if p_tag:
                print(f"Iframe #{idx}, Div #{div_idx}: {p_tag.get_text()}")
            else:
                print(f"Iframe #{idx}, Div #{div_idx}: No <p> tag with the specified class found.")
    else:
        print(f"Iframe #{idx}: No src attribute found.")
```
